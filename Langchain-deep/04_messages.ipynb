# Messages
- Messagess are the fundamentals unit of context for models in Langchain. They represent the input and output of models, carrying both the content and metadata needed to represent the state of a conversation when interacting with an LLM. Messages are objects that contain:

 1. Role - Identifies the messages type (e.g. System, user)
 2. Content - Represents the actual content of the message (like, text,images, audio, documents,. etc)
 3. Metadata - Optional fields such as response information, message ID's, and token usage

Langchain provides a standard message type that works across all model providers, ensuring consistent behavior regardless of the model being called. 

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
model = init_chat_model("groq:openai/gpt-oss-20b")
```

```python
model.invoke("please tell what is artifical intelligenece in 3 bullet points")
```

# TExt prompts
- Text prompts are strings - ideal for straightforward generation tasks where you don't need to retain conversation history
- use text prompts when: 
 1. you have a single, standalone request
 2. You dont need conversation history
 3. You want minimal code complexity

```python
model.invoke("What is langchain")
```

# Message prompts
 - Alternatively , you can pass in a list of messages to the model by providing a list of message objects

Message Prompts
1. System message - TElls the model how to behave and provide context for interactions
2. Human Message - Represents user input and interacting with the model
3. AI message - Responses generated by the model, including text content, tool calls, and metadata
4. Tool message - Represents the outputs of tool calls

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage("you are a poetry expert"),
    HumanMessage("write a poem on artificial intelligence"),
]

response = model.invoke(messages)
response.content
```

```python


system_msg = SystemMessage("you are a helpful coding assistant.")

messages = [
    system_msg, HumanMessage("How do i create a REST API?")
]
response = model.invoke(messages)
print(response.content)
```

```python
# detailed info to the LLM throught system message

system_msg = SystemMessage("""
You are a senior python developer with expertise in web frameworks.
Always provide code examples and explain your reasoning.
Be concise but thorought in your explanations.
""")

messages = [
    system_msg,HumanMessage("How do i create a REST API?")
]
response = model.invoke(messages)
print(response.content)
```

```python
# human message
human_msg = HumanMessage(
    content="Hello!",
    name = "kumar",
    id = "msg_123"
)

response = model.invoke([human_msg])
response
```

```python
from langchain.messages import AIMessage, SystemMessage, HumanMessage

# create an AI agent manually. for converstaion history
ai_msg = AIMessage("I'd be happy to help you with that question!")

# add conversation history
messages = [
    SystemMessage("You are a helpful assistant."),
    HumanMessage("Can you help me?"),
    ai_msg,
    HumanMessage("Great! What's 2x2?")
]

response = model.invoke(messages)
print(response.content)
```

# Tool message

```python
from langchain.messages import AIMessage,HumanMessage, SystemMessage
from langchain.messages import ToolMessage

# after the model makes a tool call
# (Here, we demonstrate manually creating the messages for brevity)

ai_message = AIMessage(
    content=[],
    tool_calls=[{
        "name":"get_weather",
        "args":{"location":"Tamil Nadu"},
        "id": "call_123"
    }]
)

# Execute tool and create result message

weather_result = "Sunny, 72 degree"
tool_message = ToolMessage(
    content=weather_result,
    tool_call_id ="call_123" # must match the call ID
)

# continue conversation
messages = [
    HumanMessage("What's the weather in Tamil Nadu?"),
    ai_message,
    tool_message
]

response = model.invoke(messages)
response
```
